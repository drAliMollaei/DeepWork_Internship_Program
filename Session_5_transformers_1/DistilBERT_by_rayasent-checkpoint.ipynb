```python

import pandas as pd

# Load the dataset
df = pd.read_excel('RayaSent_balanced_data.xlsx', header=None)
df.columns = ['comment', 'rate']
print(df.shape)
df.head(3)
```

```python
# Preprocess the dataset
df['label'] = df['rate'].map({'Extremely bad':1,'Bad':2,'Average':3, 'Good':4, 'Excellent':5 }) # Convert labels to integers
df['text'] = df['comment'].astype(str) # Ensure all data are strings
df.head(3)
```

```python
from sklearn.model_selection import train_test_split

# Split the data into training and testing sets
train_texts, test_texts, train_labels, test_labels = train_test_split(df['text'].tolist(), df['label'].tolist(), test_size=0.2)
```

```python
# !pip install setuptools
!pip install --upgrade packaging
```

```python
from transformers import DistilBertTokenizerFast
# Tokenize the data
# tokenizer = DistilBertTokenizerFast.from_pretrained('distilbert-base-uncased')
tokenizer_path = "./model"

# Load the tokenizer from the local directory
tokenizer = DistilBertTokenizerFast.from_pretrained(tokenizer_path, local_files_only=True)

train_encodings = tokenizer(train_texts, truncation=True, padding=True, max_length=512)
test_encodings  = tokenizer(test_texts, truncation=True, padding=True, max_length=512)
```

attention_mask: A list of 0s and 1s indicating which tokens are real words and which are padding tokens. The model focuses its attention only on real words.

```python
import torch
```

```python
class TweetDataset(torch.utils.data.Dataset):
    def __init__(self, encodings, labels):
        self.encodings = encodings
        self.labels = labels

    def __getitem__(self, idx):
        item = {key: torch.tensor(val[idx]) for key, val in self.encodings.items()}
        item['labels'] = torch.tensor(self.labels[idx])
        return item

    def __len__(self):
        return len(self.labels)
```

Inheritance: It inherits from torch.utils.data.Dataset, providing a foundation for defining custom datasets in PyTorch.
Constructor (__init__):
Takes two arguments:
encodings: The list of encoded tweet samples, as prepared by the tokenizer.
labels: The corresponding list of labels for each sample.
Stores these as attributes for later access.
Item Access (__getitem__):
Enables accessing individual samples by index.
Retrieves the encoded tensors for a specific sample from the encodings list.
Fetches the corresponding label from the labels list.
Packages them together into a dictionary for seamless model input.
Length (__len__):
Returns the total number of samples in the dataset, used for iterating and batching.

```python
train_dataset = TweetDataset(train_encodings, train_labels)
test_dataset  = TweetDataset(test_encodings, test_labels)
```

```python
# from transformers import RobertaForSequenceClassification

# # Define the model
# model = RobertaForSequenceClassification.from_pretrained('roberta-base', num_labels=4)

from transformers import DistilBertForSequenceClassification

# Define the model
model = DistilBertForSequenceClassification.from_pretrained('./', num_labels=4)
```

```python
from transformers import TrainingArguments

# Define the training arguments
training_args = TrainingArguments(
    output_dir='./results',
    num_train_epochs=1,
    per_device_train_batch_size=32,
    per_device_eval_batch_size=64,
    warmup_steps=500,
    weight_decay=0.01,
    logging_dir='./logs',
    logging_steps=10,
    load_best_model_at_end=True,
    evaluation_strategy="epoch",
    save_strategy="epoch",
    metric_for_best_model="accuracy",
)
```

```python
!pip install --upgrade setuptools
```

```python
from transformers import Trainer
from sklearn.metrics import accuracy_score

# Define the trainer
trainer = Trainer(
    model=model,
    args=training_args,
    train_dataset=train_dataset,
    eval_dataset=test_dataset,
    compute_metrics=lambda eval_pred: {'accuracy': accuracy_score(eval_pred.label_ids,
                                                                  np.argmax(eval_pred.predictions, axis=1)
                                                                  )
                                      }
)
```

```python
# Train the model
trainer.train()
```

```python
# Evaluate the model
eval_result = trainer.evaluate()

print(f"Accuracy: {eval_result['eval_accuracy']}")
```
